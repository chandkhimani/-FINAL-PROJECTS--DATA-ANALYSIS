```python
# ============================================================
# POLLUTANT + WEATHER CORRELATION HEATMAP
# ============================================================

# Create correlation matrix first
pollutant_weather_corr = df[
    pollutant_columns + weather_columns
].corr()

print("Pollutant + Weather Correlation Matrix:")
display(pollutant_weather_corr.round(2))

# Create heatmap
plt.figure(figsize=(12, 8))

sns.heatmap(
    pollutant_weather_corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Between Pollutants and Weather Variables")
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# STRONGEST POLLUTANT CORRELATIONS
# ============================================================

corr_matrix = df[pollutant_columns].corr()

correlation_pairs = []

for i in range(len(corr_matrix.columns)):
    for j in range(i + 1, len(corr_matrix.columns)):

        col1 = corr_matrix.columns[i]
        col2 = corr_matrix.columns[j]

        correlation_value = corr_matrix.iloc[i, j]

        correlation_pairs.append(
            [col1, col2, correlation_value]
        )

correlation_pairs_df = pd.DataFrame(
    correlation_pairs,
    columns=[
        'Pollutant 1',
        'Pollutant 2',
        'Correlation'
    ]
)

correlation_pairs_df['Absolute Correlation'] = (
    correlation_pairs_df['Correlation'].abs()
)

correlation_pairs_df = (
    correlation_pairs_df
    .sort_values(
        'Absolute Correlation',
        ascending=False
    )
)

print("Pollutant Correlation Pairs:")

display(
    correlation_pairs_df.round(3)
)
```

```python
# ============================================================
# WEATHER-POLLUTANT CORRELATIONS
# ============================================================

weather_pollutant_results = []

for weather in weather_columns:

    for pollutant in pollutant_columns:

        correlation = df[weather].corr(
            df[pollutant]
        )

        weather_pollutant_results.append(
            [weather, pollutant, correlation]
        )

weather_pollutant_df = pd.DataFrame(
    weather_pollutant_results,
    columns=[
        'Weather Variable',
        'Pollutant',
        'Correlation'
    ]
)

weather_pollutant_df['Absolute Correlation'] = (
    weather_pollutant_df['Correlation'].abs()
)

weather_pollutant_df = (
    weather_pollutant_df
    .sort_values(
        'Absolute Correlation',
        ascending=False
    )
)

print("Weather vs Pollutant Correlations:")

display(
    weather_pollutant_df.round(3)
)
```

```python
# Automatically identify strongest weather-pollutant relationship

strongest_weather_relationship = (
    weather_pollutant_df.iloc[0]
)

print("Strongest Weather-Pollutant Relationship:")

print(
    f"Weather Variable: "
    f"{strongest_weather_relationship['Weather Variable']}"
)

print(
    f"Pollutant: "
    f"{strongest_weather_relationship['Pollutant']}"
)

print(
    f"Correlation: "
    f"{strongest_weather_relationship['Correlation']:.3f}"
)
```

```python
# ============================================================
# STRONGEST WEATHER-POLLUTANT RELATIONSHIP
# ============================================================

weather_relationships = []

for weather in weather_columns:

    for pollutant in pollutant_columns:

        correlation = df[weather].corr(
            df[pollutant]
        )

        weather_relationships.append(
            [
                weather,
                pollutant,
                correlation
            ]
        )

weather_relationships_df = pd.DataFrame(
    weather_relationships,
    columns=[
        'Weather Variable',
        'Pollutant',
        'Correlation'
    ]
)

weather_relationships_df['Absolute Correlation'] = (
    weather_relationships_df['Correlation'].abs()
)

weather_relationships_df = (
    weather_relationships_df
    .sort_values(
        'Absolute Correlation',
        ascending=False
    )
    .reset_index(drop=True)
)

print("Weather-Pollutant Correlations:")

display(
    weather_relationships_df.round(3)
)
```

```python
# ============================================================
# STRONGEST POLLUTANT CORRELATION
# ============================================================

pollutant_corr = df[pollutant_columns].corr()

pollutant_pairs = []

for i in range(len(pollutant_corr.columns)):

    for j in range(i + 1, len(pollutant_corr.columns)):

        pollutant_1 = pollutant_corr.columns[i]
        pollutant_2 = pollutant_corr.columns[j]

        correlation = pollutant_corr.iloc[i, j]

        pollutant_pairs.append(
            [
                pollutant_1,
                pollutant_2,
                correlation
            ]
        )

pollutant_pairs_df = pd.DataFrame(
    pollutant_pairs,
    columns=[
        'Pollutant 1',
        'Pollutant 2',
        'Correlation'
    ]
)

pollutant_pairs_df['Absolute Correlation'] = (
    pollutant_pairs_df['Correlation'].abs()
)

pollutant_pairs_df = (
    pollutant_pairs_df
    .sort_values(
        'Absolute Correlation',
        ascending=False
    )
    .reset_index(drop=True)
)

top_pollutant_pair = pollutant_pairs_df.iloc[0]

print("Strongest Pollutant-Pollutant Relationship:")
print(
    f"{top_pollutant_pair['Pollutant 1']} "
    f"and "
    f"{top_pollutant_pair['Pollutant 2']}"
)

print(
    f"Correlation: "
    f"{top_pollutant_pair['Correlation']:.3f}"
)
```
