```python
# ============================================================
# 🧹 STANDARDIZE COLUMN NAMES
# ============================================================

def clean_column_names(df):
    """
    Standardize column names by:
    - Removing leading/trailing spaces
    - Converting to lowercase
    - Replacing spaces with underscores
    - Removing special characters
    """

    df = df.copy()

    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(r"[^a-z0-9]+", "_", regex=True)
        .str.strip("_")
    )

    return df


# Apply standardization to every yearly dataset
cleaned_data = {}

for year, df in happiness_data.items():
    cleaned_data[year] = clean_column_names(df)

print("✅ Column names standardized successfully!\n")

for year, df in cleaned_data.items():
    print(f"📅 {year}:")
    print(list(df.columns))
    print()
```

```python
# ============================================================
# 🔎 IDENTIFY KEY HAPPINESS VARIABLES
# ============================================================

keywords = [
    "happiness",
    "score",
    "gdp",
    "economy",
    "social",
    "support",
    "health",
    "life",
    "expectancy",
    "freedom",
    "generosity",
    "corruption"
]

for year, df in cleaned_data.items():

    print("=" * 80)
    print(f"📅 {year} — POTENTIALLY RELEVANT COLUMNS")
    print("=" * 80)

    relevant_columns = [
        column for column in df.columns
        if any(keyword in column for keyword in keywords)
    ]

    for column in relevant_columns:
        print("•", column)
```

```python
# ============================================================
# 🌍 COUNTRY COLUMN IDENTIFICATION
# ============================================================

for year, df in cleaned_data.items():

    country_columns = [
        column for column in df.columns
        if any(keyword in column for keyword in ["country", "region", "name"])
    ]

    print(f"📅 {year} → {country_columns}")
```

```python
# ============================================================
# 🔄 STANDARDIZE VARIABLES ACROSS ALL YEARS
# ============================================================

def standardize_happiness_data(df, year):
    """
    Convert year-specific World Happiness Report columns
    into a common structure for multi-year analysis.
    """

    df = df.copy()

    # Column mapping for different dataset versions
    if year <= 2017:
        column_mapping = {
            "country": "country",
            "happiness_score": "happiness_score",
            "economy_gdp_per_capita": "gdp_per_capita",
            "family": "social_support",
            "health_life_expectancy": "life_expectancy",
            "freedom": "freedom",
            "generosity": "generosity",
            "trust_government_corruption": "corruption_perception"
        }

    else:
        column_mapping = {
            "country_or_region": "country",
            "score": "happiness_score",
            "gdp_per_capita": "gdp_per_capita",
            "social_support": "social_support",
            "healthy_life_expectancy": "life_expectancy",
            "freedom_to_make_life_choices": "freedom",
            "generosity": "generosity",
            "perceptions_of_corruption": "corruption_perception"
        }

    # Rename only columns that actually exist
    available_mapping = {
        old: new
        for old, new in column_mapping.items()
        if old in df.columns
    }

    df = df.rename(columns=available_mapping)

    # Keep only standardized analytical variables
    required_columns = [
        "country",
        "happiness_score",
        "gdp_per_capita",
        "social_support",
        "life_expectancy",
        "freedom",
        "generosity",
        "corruption_perception"
    ]

    # Create missing columns if necessary
    for column in required_columns:
        if column not in df.columns:
            df[column] = np.nan

    df = df[required_columns]

    # Add year information
    df["year"] = year

    return df


# Apply standardization to every year
standardized_data = {}

for year, df in cleaned_data.items():
    standardized_data[year] = standardize_happiness_data(df, year)

print("✅ All datasets standardized successfully!")
```

```python
# ============================================================
# 🌍 CREATE MASTER DATASET
# ============================================================

master_df = pd.concat(
    standardized_data.values(),
    ignore_index=True
)

# Arrange columns logically
master_df = master_df[
    [
        "year",
        "country",
        "happiness_score",
        "gdp_per_capita",
        "social_support",
        "life_expectancy",
        "freedom",
        "generosity",
        "corruption_perception"
    ]
]

print("=" * 80)
print("🌍 MASTER DATASET CREATED")
print("=" * 80)

print(f"📊 Total Records : {master_df.shape[0]}")
print(f"📋 Total Columns : {master_df.shape[1]}")

display(master_df.head(10))
```

# 📊 3. Happiness Factor Analysis

This section investigates the relationship between happiness and major
economic, social, health, and well-being indicators.

The analysis focuses on six key explanatory variables:

- 💰 GDP per Capita
- 🤝 Social Support
- ❤️ Healthy Life Expectancy
- 🕊️ Freedom
- 🎁 Generosity
- 🏛️ Corruption Perception

Both visual and statistical techniques are used to evaluate the strength
and direction of these relationships.

```python
# ============================================================
# 📊 HAPPINESS SCORE DISTRIBUTION
# ============================================================

plt.figure(figsize=(11, 6))

sns.histplot(
    data=happiness_df,
    x="happiness_score",
    bins=25,
    kde=True
)

plt.title(
    "Distribution of Happiness Scores (2015–2019)",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Happiness Score", fontsize=12)
plt.ylabel("Number of Country-Year Records", fontsize=12)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 🔗 CORRELATION MATRIX
# ============================================================

factor_columns = [
    "happiness_score",
    "gdp_per_capita",
    "social_support",
    "life_expectancy",
    "freedom",
    "generosity",
    "corruption_perception"
]

correlation_matrix = happiness_df[factor_columns].corr()

display(correlation_matrix.round(3))
```

```python
# ============================================================
# 🔥 CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    linewidths=0.5,
    square=True
)

plt.title(
    "Correlation Matrix of Happiness and Key Factors",
    fontsize=17,
    fontweight="bold",
    pad=15
)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 🏆 FACTOR CORRELATION RANKING
# ============================================================

factor_correlations = (
    correlation_matrix["happiness_score"]
    .drop("happiness_score")
    .sort_values(ascending=False)
)

correlation_ranking = (
    factor_correlations
    .to_frame(name="Correlation with Happiness")
)

correlation_ranking["Absolute Correlation"] = (
    correlation_ranking["Correlation with Happiness"]
    .abs()
)

correlation_ranking["Strength Rank"] = (
    correlation_ranking["Absolute Correlation"]
    .rank(ascending=False, method="min")
    .astype(int)
)

display(correlation_ranking.round(3))
```

```python
# ============================================================
# 📊 FACTOR CORRELATION RANKING VISUALIZATION
# ============================================================

ranking_plot = (
    factor_correlations
    .sort_values()
    .reset_index()
)

ranking_plot.columns = [
    "Factor",
    "Correlation"
]

plt.figure(figsize=(11, 6))

sns.barplot(
    data=ranking_plot,
    x="Correlation",
    y="Factor"
)

plt.axvline(0, linewidth=1)

plt.title(
    "Strength of Association with Happiness Score",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Pearson Correlation", fontsize=12)
plt.ylabel("Factor", fontsize=12)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 💰 GDP PER CAPITA vs HAPPINESS
# ============================================================

gdp_corr = happiness_df[
    ["gdp_per_capita", "happiness_score"]
].corr().iloc[0, 1]

plt.figure(figsize=(11, 7))

sns.regplot(
    data=happiness_df,
    x="gdp_per_capita",
    y="happiness_score",
    scatter_kws={"alpha": 0.55},
    line_kws={"linewidth": 2}
)

plt.title(
    f"GDP per Capita vs Happiness Score\n"
    f"Pearson Correlation = {gdp_corr:.2f}",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("GDP per Capita", fontsize=12)
plt.ylabel("Happiness Score", fontsize=12)

plt.tight_layout()
plt.show()
```

# 📈 4. Year-wise & Country-level Comparative Analysis

This section examines how happiness and its associated factors changed
over the 2015–2019 period.

The analysis compares yearly averages, identifies countries with the
largest changes in happiness, and investigates consistency in country-level
performance over time.

```python
# ============================================================
# 📅 YEAR-WISE AVERAGE OF KEY FACTORS
# ============================================================

yearly_factors = (
    happiness_df
    .groupby("year")[
        [
            "happiness_score",
            "gdp_per_capita",
            "social_support",
            "life_expectancy",
            "freedom",
            "generosity",
            "corruption_perception"
        ]
    ]
    .mean()
    .round(3)
)

display(yearly_factors)
```

```python
# ============================================================
# 📈 YEAR-WISE HAPPINESS TREND
# ============================================================

plt.figure(figsize=(11, 6))

sns.lineplot(
    data=yearly_factors.reset_index(),
    x="year",
    y="happiness_score",
    marker="o",
    linewidth=2.5
)

plt.title(
    "Average Happiness Score by Year (2015–2019)",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Year")
plt.ylabel("Average Happiness Score")

plt.xticks(yearly_factors.index)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 📊 STANDARDIZED FACTOR TRENDS
# ============================================================

from sklearn.preprocessing import StandardScaler

factor_columns = [
    "happiness_score",
    "gdp_per_capita",
    "social_support",
    "life_expectancy",
    "freedom",
    "generosity",
    "corruption_perception"
]

yearly_factor_values = yearly_factors[factor_columns]

scaler = StandardScaler()

standardized_yearly = pd.DataFrame(
    scaler.fit_transform(yearly_factor_values),
    index=yearly_factor_values.index,
    columns=yearly_factor_values.columns
)

display(standardized_yearly.round(2))
```

```python
# ============================================================
# 📈 COMPARATIVE FACTOR TRENDS
# ============================================================

plt.figure(figsize=(13, 7))

for column in standardized_yearly.columns:
    plt.plot(
        standardized_yearly.index,
        standardized_yearly[column],
        marker="o",
        linewidth=2,
        label=column.replace("_", " ").title()
    )

plt.axhline(0, linewidth=1)

plt.title(
    "Standardized Trends of Happiness-Related Factors (2015–2019)",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Year")
plt.ylabel("Standardized Average Value")

plt.xticks(standardized_yearly.index)
plt.legend(bbox_to_anchor=(1.02, 1), loc="upper left")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 🌍 COUNTRY-LEVEL HAPPINESS SUMMARY
# ============================================================

country_summary = (
    happiness_df
    .groupby("country")
    .agg(
        Average_Happiness=("happiness_score", "mean"),
        Highest_Happiness=("happiness_score", "max"),
        Lowest_Happiness=("happiness_score", "min"),
        Years_Observed=("year", "nunique")
    )
    .sort_values(
        "Average_Happiness",
        ascending=False
    )
)

display(country_summary.head(15).round(3))
```

```python
# ============================================================
# 🏆 TOP 10 COUNTRIES BY AVERAGE HAPPINESS
# ============================================================

top_countries = (
    country_summary
    .head(10)
    .sort_values("Average_Happiness")
)

plt.figure(figsize=(11, 7))

sns.barplot(
    data=top_countries.reset_index(),
    x="Average_Happiness",
    y="country"
)

plt.title(
    "Top 10 Countries by Average Happiness (2015–2019)",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Average Happiness Score")
plt.ylabel("Country")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 📉 BOTTOM 10 COUNTRIES BY AVERAGE HAPPINESS
# ============================================================

bottom_countries = (
    country_summary
    .tail(10)
    .sort_values("Average_Happiness")
)

plt.figure(figsize=(11, 7))

sns.barplot(
    data=bottom_countries.reset_index(),
    x="Average_Happiness",
    y="country"
)

plt.title(
    "Bottom 10 Countries by Average Happiness (2015–2019)",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Average Happiness Score")
plt.ylabel("Country")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 📈 HAPPINESS CHANGE OVER TIME
# ============================================================

country_year = (
    happiness_df
    .groupby(["country", "year"])["happiness_score"]
    .mean()
    .reset_index()
)

first_year = (
    country_year
    .sort_values("year")
    .groupby("country")
    .first()
    .reset_index()
)

last_year = (
    country_year
    .sort_values("year")
    .groupby("country")
    .last()
    .reset_index()
)

country_change = first_year[
    ["country", "year", "happiness_score"]
].merge(
    last_year[
        ["country", "year", "happiness_score"]
    ],
    on="country",
    suffixes=("_first", "_last")
)

country_change["happiness_change"] = (
    country_change["happiness_score_last"]
    - country_change["happiness_score_first"]
)

country_change = country_change.sort_values(
    "happiness_change",
    ascending=False
)

display(
    country_change[
        [
            "country",
            "year_first",
            "year_last",
            "happiness_score_first",
            "happiness_score_last",
            "happiness_change"
        ]
    ].head(15).round(3)
)
```

```python
# ============================================================
# 🚀 COUNTRIES WITH LARGEST HAPPINESS IMPROVEMENT
# ============================================================

largest_improvements = (
    country_change
    .sort_values("happiness_change", ascending=False)
    .head(10)
)

display(
    largest_improvements[
        [
            "country",
            "year_first",
            "year_last",
            "happiness_change"
        ]
    ].round(3)
)
```

```python
# ============================================================
# 📉 COUNTRIES WITH LARGEST HAPPINESS DECLINE
# ============================================================

largest_declines = (
    country_change
    .sort_values("happiness_change")
    .head(10)
)

display(
    largest_declines[
        [
            "country",
            "year_first",
            "year_last",
            "happiness_change"
        ]
    ].round(3)
)
```

# 🧠 5. Advanced Statistical Analysis

Correlation analysis helps quantify the strength and direction of the
linear relationship between happiness and the selected socio-economic
and well-being indicators.

This section extends the analysis by:

- Ranking factors according to their correlation with happiness.
- Comparing relationships across individual years.
- Identifying factors with consistently strong associations.
- Distinguishing between positive and negative relationships.

> Correlation measures association, not causation.

```python
# ============================================================
# 🔗 OVERALL CORRELATION SUMMARY
# ============================================================

overall_correlation = (
    happiness_df[factor_columns]
    .corr()["happiness_score"]
    .drop("happiness_score")
    .sort_values(ascending=False)
)

correlation_summary = pd.DataFrame({
    "Factor": overall_correlation.index,
    "Correlation": overall_correlation.values
})

correlation_summary["Absolute_Correlation"] = (
    correlation_summary["Correlation"].abs()
)

correlation_summary["Relationship"] = np.where(
    correlation_summary["Correlation"] > 0,
    "Positive",
    "Negative"
)

correlation_summary = correlation_summary.sort_values(
    "Absolute_Correlation",
    ascending=False
).reset_index(drop=True)

correlation_summary.index += 1

display(correlation_summary.round(3))
```

```python
# ============================================================
# 📅 YEAR-WISE CORRELATION WITH HAPPINESS
# ============================================================

yearly_correlations = {}

for year, group in happiness_df.groupby("year"):

    correlations = (
        group[factor_columns]
        .corr()["happiness_score"]
        .drop("happiness_score")
    )

    yearly_correlations[year] = correlations

yearly_correlation_df = pd.DataFrame(yearly_correlations).T

display(yearly_correlation_df.round(3))
```

```python
# ============================================================
# 🔥 YEAR-WISE CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(12, 7))

sns.heatmap(
    yearly_correlation_df,
    annot=True,
    fmt=".2f",
    linewidths=0.5,
    center=0
)

plt.title(
    "Year-wise Correlation of Happiness with Key Factors",
    fontsize=16,
    fontweight="bold",
    pad=15
)

plt.xlabel("Factor")
plt.ylabel("Year")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 📊 CORRELATION STABILITY ACROSS YEARS
# ============================================================

correlation_stability = pd.DataFrame({
    "Average_Correlation": yearly_correlation_df.mean(),
    "Minimum_Correlation": yearly_correlation_df.min(),
    "Maximum_Correlation": yearly_correlation_df.max(),
    "Correlation_STD": yearly_correlation_df.std()
})

correlation_stability["Absolute_Average_Correlation"] = (
    correlation_stability["Average_Correlation"].abs()
)

correlation_stability = correlation_stability.sort_values(
    "Absolute_Average_Correlation",
    ascending=False
)

display(correlation_stability.round(3))
```

```python
# ============================================================
# 🏆 STRONGEST AND WEAKEST ASSOCIATIONS
# ============================================================

strongest_factor = correlation_summary.iloc[0]
weakest_factor = correlation_summary.iloc[-1]

print("=" * 80)
print("🏆 STRONGEST ASSOCIATION WITH HAPPINESS")
print("=" * 80)

print(
    f"Factor       : {strongest_factor['Factor']}"
)

print(
    f"Correlation  : {strongest_factor['Correlation']:.3f}"
)

print(
    f"Relationship : {strongest_factor['Relationship']}"
)

print("\n" + "=" * 80)
print("📉 WEAKEST ASSOCIATION WITH HAPPINESS")
print("=" * 80)

print(
    f"Factor       : {weakest_factor['Factor']}"
)

print(
    f"Correlation  : {weakest_factor['Correlation']:.3f}"
)

print(
    f"Relationship : {weakest_factor['Relationship']}"
)
```

```python
# ============================================================
# 💡 FACTOR INSIGHT SUMMARY
# ============================================================

insight_table = correlation_summary[
    ["Factor", "Correlation", "Relationship"]
].copy()

insight_table["Interpretation"] = insight_table.apply(
    lambda row:
        f"{'Positive' if row['Correlation'] > 0 else 'Negative'} "
        f"association with happiness "
        f"(r = {row['Correlation']:.2f})",
    axis=1
)

display(insight_table.round(3))
```
