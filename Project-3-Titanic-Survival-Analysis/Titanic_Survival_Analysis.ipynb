```python
# ============================================
# 🛠️ HANDLE MISSING AGE
# ============================================

if "Age" in df.columns:
    age_median = df["Age"].median()

    df["Age"] = df["Age"].fillna(age_median)

    print(f"✅ Missing Age values filled with median: {age_median:.2f}")
    print("Remaining missing Age values:", df["Age"].isnull().sum())
```

```python
# ============================================
# 🚢 EMBARKED ANALYSIS
# ============================================

if "Embarked" in df.columns:
    print("Missing Embarked Values:",
          df["Embarked"].isnull().sum())

    print("\nEmbarked Value Counts:")
    display(df["Embarked"].value_counts(dropna=False))
```

```python
# ============================================
# 🛠️ HANDLE MISSING EMBARKED
# ============================================

if "Embarked" in df.columns:

    if df["Embarked"].isnull().sum() > 0:

        embarked_mode = df["Embarked"].mode()[0]

        df["Embarked"] = df["Embarked"].fillna(embarked_mode)

        print(f"✅ Missing Embarked values filled with mode: {embarked_mode}")

    else:
        print("✅ No missing Embarked values found.")

    print(
        "Remaining missing Embarked values:",
        df["Embarked"].isnull().sum()
    )
```

```python
# ============================================
# 🛏️ CABIN ANALYSIS
# ============================================

if "Cabin" in df.columns:

    cabin_missing = df["Cabin"].isnull().sum()
    cabin_percentage = cabin_missing / len(df) * 100

    print("Missing Cabin Values:", cabin_missing)
    print(f"Missing Cabin Percentage: {cabin_percentage:.2f}%")
```

```python
# ============================================
# 🛏️ CABIN AVAILABILITY
# ============================================

if "Cabin" in df.columns:

    df["CabinKnown"] = df["Cabin"].notna().astype(int)

    print("✅ CabinKnown feature created.")

    display(
        df[["Cabin", "CabinKnown"]].head()
    )
```

```python
# ============================================
# 👨‍👩‍👧 FAMILY SIZE
# ============================================

if "SibSp" in df.columns and "Parch" in df.columns:

    df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

    print("✅ FamilySize feature created.")

    display(
        df[["SibSp", "Parch", "FamilySize"]].head()
    )
```

```python
# ============================================
# 🧳 TRAVELLING ALONE
# ============================================

if "FamilySize" in df.columns:

    df["IsAlone"] = np.where(
        df["FamilySize"] == 1,
        1,
        0
    )

    print("✅ IsAlone feature created.")

    display(
        df[["FamilySize", "IsAlone"]].head()
    )
```

```python
# ============================================
# 🎂 AGE GROUP
# ============================================

if "Age" in df.columns:

    def create_age_group(age):

        if age < 13:
            return "Child"

        elif age < 20:
            return "Teen"

        elif age < 60:
            return "Adult"

        else:
            return "Senior"

    df["AgeGroup"] = df["Age"].apply(create_age_group)

    print("✅ AgeGroup feature created.")

    display(
        df[["Age", "AgeGroup"]].head(10)
    )
```

```python
# ============================================
# 🔍 FINAL MISSING VALUE CHECK
# ============================================

print("### Missing Values After Cleaning")

final_missing = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

display(
    final_missing[
        final_missing["Missing Count"] > 0
    ]
)
```

```python
# ============================================
# 🔁 FINAL DUPLICATE CHECK
# ============================================

print("Duplicate Rows After Cleaning:",
      df.duplicated().sum())
```

```python
# ============================================
# 📐 FINAL DATASET SHAPE
# ============================================

print("Final Dataset Shape:")
print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])
```

## 🧹 Data Cleaning Decisions

The dataset was checked for missing values, duplicate rows, incorrect
data types, and other potential data-quality issues.

### Age
Missing Age values were handled using median imputation because age
is an important numerical variable and median provides a simple,
robust approach.

### Embarked
Missing Embarked values, if present, were handled using the mode
(the most frequently occurring category).

### Cabin
Cabin information was inspected for missingness. Instead of
fabricating cabin values, a `CabinKnown` feature was created to
indicate whether cabin information was available.

### Duplicate Rows
Duplicate rows were checked before further analysis.

### Feature Creation
Three simple analytical features were created:

- `FamilySize` = SibSp + Parch + 1
- `IsAlone` = 1 when FamilySize is 1, otherwise 0
- `AgeGroup` = Child, Teen, Adult, or Senior

These features will help analyze family and age-related survival
patterns in the next sections.

### ⚠️ Understanding Correlation

Correlation measures the strength and direction of association between
numerical variables.

A positive correlation means that two variables tend to increase
together, while a negative correlation means that one tends to
increase as the other decreases.

However, correlation does NOT prove causation.

Therefore, the correlation results in this project should be treated
as indicators of association rather than proof that one variable
directly caused another.

```python
# ============================================
# 👩‍🦰 GENDER-BASED INSIGHT
# ============================================

gender_rates = (
    df.groupby("Sex")["Survived"]
    .mean()
    .mul(100)
    .round(2)
)

highest_gender = gender_rates.idxmax()
highest_gender_rate = gender_rates.max()

lowest_gender = gender_rates.idxmin()
lowest_gender_rate = gender_rates.min()

print("👩‍🦰 GENDER INSIGHT")
print("=" * 50)

for gender, rate in gender_rates.items():
    print(f"{gender.title()} Survival Rate: {rate:.2f}%")

print(f"\nHighest survival rate: {highest_gender.title()} ({highest_gender_rate:.2f}%)")
print(f"Lowest survival rate: {lowest_gender.title()} ({lowest_gender_rate:.2f}%)")
```

## 👩‍🦰 Gender-Based Finding

The gender analysis shows a clear difference in survival rates between
male and female passengers.

The calculated survival rates above are based directly on the dataset.

```python
# ============================================
# 🎫 PASSENGER CLASS INSIGHT
# ============================================

class_rates = (
    df.groupby("Pclass")["Survived"]
    .mean()
    .mul(100)
    .round(2)
)

highest_class = class_rates.idxmax()
highest_class_rate = class_rates.max()

lowest_class = class_rates.idxmin()
lowest_class_rate = class_rates.min()

print("🎫 PASSENGER CLASS INSIGHT")
print("=" * 50)

for pclass, rate in class_rates.items():
    print(f"Class {pclass} Survival Rate: {rate:.2f}%")

print(
    f"\nHighest survival rate: Class {highest_class} "
    f"({highest_class_rate:.2f}%)"
)

print(
    f"Lowest survival rate: Class {lowest_class} "
    f"({lowest_class_rate:.2f}%)"
)
```

```python
# ============================================
# 🎂 AGE GROUP INSIGHT
# ============================================

age_rates = (
    df.groupby("AgeGroup")["Survived"]
    .mean()
    .mul(100)
    .round(2)
    .sort_values(ascending=False)
)

print("🎂 AGE GROUP INSIGHT")
print("=" * 50)

display(
    age_rates.reset_index(
        name="Survival Rate (%)"
    )
)

highest_age_group = age_rates.idxmax()
highest_age_rate = age_rates.max()

print(
    f"Highest survival rate among age groups: "
    f"{highest_age_group} ({highest_age_rate:.2f}%)"
)
```

## 🎂 Age-Based Finding

Survival rates were compared across the age groups created during the
data transformation stage.

This helps identify whether survival outcomes differed across age
categories.

```python
# ============================================
# 👨‍👩‍👧 FAMILY SIZE INSIGHT
# ============================================

family_rates = (
    df.groupby("FamilySize")["Survived"]
    .mean()
    .mul(100)
    .round(2)
    .sort_values(ascending=False)
)

highest_family_size = family_rates.idxmax()
highest_family_rate = family_rates.max()

lowest_family_size = family_rates.idxmin()
lowest_family_rate = family_rates.min()

print("👨‍👩‍👧 FAMILY SIZE INSIGHT")
print("=" * 50)

display(
    family_rates.reset_index(
        name="Survival Rate (%)"
    )
)

print(
    f"Highest survival rate: Family Size {highest_family_size} "
    f"({highest_family_rate:.2f}%)"
)

print(
    f"Lowest survival rate: Family Size {lowest_family_size} "
    f"({lowest_family_rate:.2f}%)"
)
```

```python
# ============================================
# 💰 FARE INSIGHT
# ============================================

average_fare = (
    df.groupby("Survived")["Fare"]
    .mean()
    .round(2)
)

survived_fare = average_fare.get(1)
not_survived_fare = average_fare.get(0)

print("💰 FARE INSIGHT")
print("=" * 50)

print(f"Average Fare of Survivors     : {survived_fare:.2f}")
print(f"Average Fare of Non-Survivors : {not_survived_fare:.2f}")

if survived_fare > not_survived_fare:
    print("\nSurvivors had a higher average fare.")
elif survived_fare < not_survived_fare:
    print("\nNon-survivors had a higher average fare.")
else:
    print("\nBoth groups had the same average fare.")
```

```python
# ============================================
# 🚢 EMBARKATION INSIGHT
# ============================================

port_rates = (
    df.groupby("Embarked")["Survived"]
    .mean()
    .mul(100)
    .round(2)
    .sort_values(ascending=False)
)

highest_port = port_rates.idxmax()
highest_port_rate = port_rates.max()

print("🚢 EMBARKATION INSIGHT")
print("=" * 50)

display(
    port_rates.reset_index(
        name="Survival Rate (%)"
    )
)

print(
    f"Highest survival rate by embarkation port: "
    f"{highest_port} ({highest_port_rate:.2f}%)"
)
```

```python
# ============================================
# 🧳 ALONE VS FAMILY INSIGHT
# ============================================

travel_rates = (
    df.groupby("IsAlone")["Survived"]
    .mean()
    .mul(100)
    .round(2)
)

# Convert 0 and 1 into readable labels
travel_rates.index = travel_rates.index.map({
    0: "With Family",
    1: "Alone"
})

print("🧳 TRAVEL TYPE INSIGHT")
print("=" * 50)

display(
    travel_rates.reset_index(
        name="Survival Rate (%)"
    )
)

# Find highest survival group
highest_travel_type = travel_rates.idxmax()
highest_travel_rate = travel_rates.max()

print(
    f"Highest survival rate: {highest_travel_type} "
    f"({highest_travel_rate:.2f}%)"
)
```
